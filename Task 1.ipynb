Build a machine learning model to predict the aggregate rating of a restaurant based on other features.

```python
# import libraries
import pandas as pd
import numpy as np

# load dataframe
df = pd.read_csv("C:/Users/User/Downloads/Dataset.csv"); df
```

# Data cleaning

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# fill missing values in categorical columns
df['Cuisines'] = df['Cuisines'].fillna('Unknown')
df['Rating text'] = df['Rating text'].fillna('Not Rated')

# Remove rows only if the target variable is missing
df = df.dropna(subset=['Aggregate rating'])
```

```python
df.isnull().sum()
```

# Data preprocessing and splitting

```python
# encode categorical variables into numerical values
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Has Table booking'] = le.fit_transform(df['Has Table booking'])
df['Has Online delivery'] = le.fit_transform(df['Has Online delivery'])
df['Is delivering now'] = le.fit_transform(df['Is delivering now'])
df['Switch to order menu'] = le.fit_transform(df['Switch to order menu'])
df['Rating color'] = le.fit_transform(df['Rating color'])
df['Rating text'] = le.fit_transform(df['Rating text'])
df['City'] = le.fit_transform(df['City'])
df['Currency'] = le.fit_transform(df['Currency'])
print(df)
```

```python
# select features and target
x = df[['Average Cost for two', 'Votes', 'Price range', 'Has Online delivery', 'Has Table booking']]; x
y = df['Aggregate rating']; y
```

```python
# split dataset into training and testing data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

print(x_train)
print(x_test)
print(y_train)
print(y_test)
```

# Linear Regression Model

```python
# Linear Regression Algorithm
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
print(y_pred)
```

```python
# evaluate the Linear Regression model using Mean Squared Error and R² Score
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

# Decision Tree

```python
# decision tree regressor algorithm
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state=42)
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
print(y_pred)
```

```python
# evaluate the Decision Tree model using Mean Squared Error and R² Score
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

# Feature Importance Analysis

```python
importance = pd.DataFrame({'Feature': x.columns, 'Importance': model.feature_importances_})
importance = importance.sort_values(by='Importance', ascending=False)

print(importance)
```

The feature importance analysis shows that Price range is the most influential feature affecting restaurant ratings, followed by the remaining features. This indicates that a restaurant's pricing plays an important role in predicting its aggregate rating.
