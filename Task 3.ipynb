# Data Preprocessing and Splitting

```python
# check Missing Values
df.isnull().sum()
```

```python
# handle Missing Values
df['Cuisines'].fillna(df['Cuisines'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
# encode categorical variables
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Cuisines'] = le.fit_transform(df['Cuisines'])
df['Has Table booking'] = le.fit_transform(df['Has Table booking'])
df['Has Online delivery'] = le.fit_transform(df['Has Online delivery'])
print(df)
```

```python
# select features and target
x = df[['Price range', 'Votes', 'Aggregate rating']]; x
y = df['Cuisines']; y

print(x.head())
print(y.head())
```

```python
# split dataset into training and testing data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train)
print(x_test)
print(y_train)
print(y_test)

print(x_train.shape)
print(x_train.dtypes)
```

```python
# Train Random Forest Model
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200, random_state=42); model
model.fit(x_train, y_train)
```

```python
# Predict Test Data
y_pred = model.predict(x_test); y_pred
```

```python
# Evaluate Model
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix)

accuracy = accuracy_score(y_test, y_pred); accuracy
print("Accuracy:", accuracy)

precision = precision_score(y_test, y_pred, average='weighted'); precision
print("Precision:", precision)

recall = recall_score(y_test, y_pred, average='weighted'); recall
print("Recall:", recall)

f1 = f1_score(y_test, y_pred, average='weighted'); f1
print("F1 Score:", f1)

print(classification_report(y_test, y_pred))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred); cm
print(cm)
```

```python
# Performance Analysis
# Count the number of restaurants for each cuisine
cuisine_count = df['Cuisines'].value_counts()
print(cuisine_count)
```
